```python
""" 
1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias
de cada letra en la cadena. Los espacios no deben ser considerados.
"""

def frecuencia_texto(texto): #Aqui creamos la funcion diciendo el parametro que le vamos a pasar, texto en este caso
    diccionario = {} # creamos el diccionario que luego vamos a rellenar

    for letra in texto: #creamos el bucle for para que recorra el texto que le hemos pasado
        if letra == " ": # aqui decimos que si hay un espacio en blanco que no lo cuente
            continue

        if letra in diccionario: # si la palabra ya la hemos visto 
            diccionario[letra] += 1 # le sumamos uno
        else:
            diccionario[letra] = 1 # si no hacemos que sea 1

    return diccionario #devolvemos el diccionario
```

```python
"""6. Escribe una función que calcule el factorial de un número de manera recursiva
"""

def factorial(n): #Creamos la funcion
    if n == 0 or n == 1:  # Aqui ponemos la condicion que hara que se pare
        return 1
    return n * factorial(n - 1)  # Aqui se vuelve a llamar a la funcion bajando un numero a la vez

print(factorial(10)) 
```

```python
""" 37. Crea una función llamada procesar_texto que procesa un texto según la opción especificada: contar_palabras ,
reemplazar_palabras , eliminar_palabra . Estas opciones son otras funciones que tenemos que definir primero y llamar dentro
de la función procesar_texto .
Código a seguir:
1. Crear una función contar_palabras para contar el número de veces que aparece cada palabra en el texto. Tiene
que devolver un diccionario.
2. Crear una función reemplazar_palabras para remplazar una palabra_original del texto por una palabra_nueva . Tiene
que devolver el texto con el remplazo de palabras.
3. Crear una función eliminar_palabra para eliminar una palabra del texto. Tiene que devolver el texto con la palabra
eliminada.
4. Crear la función procesar_texto que tome un texto, una opción(entre "contar", "reemplazar", "eliminar") y un
número de argumentos variable según la opción indicada.
Caso de uso:
Comprueba el funcionamiento completo de la función procesar_texto""" 


def contar_palabras(texto): # creamos las funciones
    
    conteo = {}
    for palabra in texto.split():
        conteo[palabra] = conteo.get(palabra, 0) + 1
    return conteo 


def reemplazar_palabras(texto,original, nuevo):
    return texto.replace(original, nuevo)

def eliminar_palabra(texto,palabra):
    
    palabras = texto.split()
    resultado = [p for p in palabras if p != palabra]
    
    return " ".join(resultado)

def procesar_texto(texto, opcion, *argumentos):
    
    if opcion == "contar": #aqui pondremos lo que ocurrira al elegir las diferentes opciones
        return contar_palabras(texto) # retornamos la llamada a la funcion con la opcion que toque
    
    elif opcion == "reemplazar":
        if len(argumentos) != 2:
            raise ValueError("No se puede reemplazar palabras, añada solo dos palabras")
        return reemplazar_palabras(texto, argumentos[0], argumentos[1])
    
    elif opcion == "eliminar":
        if len(argumentos) != 1:
            raise ValueError("Pasa una palabra para poder eliminar")
        return eliminar_palabra(texto, argumentos[0])
    
    else:
        raise ValueError("Opcion incorrecta")


procesar_texto("hola mundo", "eliminar", "mundo")
```

```python
""" 40. Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo" , "circulo" o
"triangulo" ) y datos (una tupla con los datos necesarios para calcular el área de la figura).""" 
import math

def calcular_area_figura(figura, datos):
    
    if figura == "rectangulo":
        base, altura = datos # dependiendo de que pongamos cambiaremos el valor de datos
        
        return base * altura
    
    elif figura == "circulo":
        radio = datos[0] # con circulos ponemos que saque el primer valor de la tupla
        
        return math.pi * radio ** 2
    
    elif figura == "triangulo":
        base, altura = datos
        
        return (base * altura) / 2

    else:
        raise ValueError("La figura que has puesto no es valida")


        
calcular_area_figura("triangulo", (6, 4))
```
